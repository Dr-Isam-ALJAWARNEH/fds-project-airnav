Below is a complete Google Colab code snippet that lets you upload your five CSV files and one GeoJSON file, merges the CSV files, reads the GeoJSON, performs a merge (using a common key if available or simply attaches a default geometry), and then converts the result into a Table abstraction using the datascience library.

What This Code Does:
File Upload:
Prompts you to upload all required files. It prints a list so you know what files to upload.

CSV Processing:
Reads each CSV file. If a file is missing (like NYC_AQ.csv), it prints a warning and continues with the rest.

GeoJSON Handling:
Reads the GeoJSON file from a temporary file, then removes that temporary file.

Merging Strategy:
Attempts to merge on a common key (if columns like polygon_id in the CSV and id in the GeoJSON exist). Otherwise, it attaches the first geometry from the GeoJSON to each CSV row.

Table Conversion:
Converts the final merged DataFrame into a datascience Table abstraction and displays it.

If you expect NYC_AQ.csv to be part of your analysis, make sure to include it when uploading your files. Otherwise, the code will simply process the ones you uploaded.

```python
import io
import tempfile
import os
import pandas as pd
import geopandas as gpd
from datascience import Table
from google.colab import files
```

```python
# Prompt the user to upload the files
print("Please upload the CSV and GeoJSON files as prompted.")
uploaded = files.upload()
```

```python
# List of expected CSV file names
expected_csv_files = [
    'NYC_AQ.csv',
    'NYC_PM.csv',
    'NYC_PM_Part1.csv',
    'NYC_PM_Part2.csv',
    'NYC_PM_Part3.csv'
]
```

```python
# Read and concatenate CSV files
df_list = []
for file in expected_csv_files:
    if file in uploaded:
        print(f"Reading {file} ...")
        df = pd.read_csv(io.BytesIO(uploaded[file]))
        df_list.append(df)
    else:
        print(f"Warning: {file} not uploaded.")

if df_list:
    df_csv = pd.concat(df_list, ignore_index=True)
else:
    raise ValueError("No CSV files were loaded.")
```

```python
# Read the GeoJSON file
geojson_filename = 'nyc_polygon.geojson'
if geojson_filename in uploaded:
    print(f"Reading {geojson_filename} ...")
    # Write the GeoJSON bytes to a temporary file so that GeoPandas can read it
    with tempfile.NamedTemporaryFile(suffix='.geojson', delete=False) as tmp:
        tmp.write(uploaded[geojson_filename])
        geojson_path = tmp.name
    gdf = gpd.read_file(geojson_path)
    os.remove(geojson_path)  # Clean up the temporary file
else:
    print(f"Warning: {geojson_filename} not uploaded.")
    gdf = None
```

```python
# Merge CSV data with GeoJSON based on a common key if available,
# otherwise attach a default geometry from the GeoJSON to each row.
if gdf is not None:
    if 'polygon_id' in df_csv.columns and 'id' in gdf.columns:
        print("Merging using common key 'polygon_id' and 'id'...")
        df_merged = pd.merge(df_csv, gdf, left_on='polygon_id', right_on='id', how='left')
    else:
        print("No common key found. Attaching a default geometry from the GeoJSON...")
        if not gdf.empty:
            df_csv['geometry'] = gdf.geometry.iloc[0]
        else:
            df_csv['geometry'] = None
        df_merged = df_csv
else:
    print("GeoJSON file was not uploaded; proceeding with CSV data only.")
    df_merged = df_csv
```

```python
# Convert the merged DataFrame to a Table abstraction from the datascience library
table = Table.from_df(df_merged)
print("Displaying the Table abstraction:")
table.show(3)
```

Below is an updated Google Colab code snippet that not only uploads, merges, and converts the files into a Table abstraction but also cleans the merged data. The cleaning process in this example:

Converts column names to lowercase for consistency

Removes duplicate rows

Drops rows with missing values (you can modify which columns are critical)

Resets the index

```python
# --- Cleaning the merged data ---
def clean_merged_data(df):
    # Convert column names to lower case for consistency
    df.columns = df.columns.str.lower()

    # Remove duplicate rows
    df = df.drop_duplicates()

    # Drop rows with missing values (modify if only specific columns are critical)
    df = df.dropna()

    # Reset the index
    df = df.reset_index(drop=True)

    return df

cleaned_df = clean_merged_data(df_merged)

# Convert the cleaned DataFrame into a Table abstraction using the datascience library
cleaned_table = Table.from_df(cleaned_df)

print("Displaying the cleaned Table abstraction:")
cleaned_table.show(3)
```

To generate geohashes from your data, you’ll need the latitude and longitude columns present in your CSV (or extracted from the GeoJSON geometry). Geohash is a great way to spatially encode locations into alphanumeric strings.

Let’s integrate geohash generation into your existing workflow using the geohash Python library. If your CSV data includes latitude and longitude columns, we’ll compute a new geohash column.

```python
# --- Geohash generation ---
!pip install geohash2

import geohash2

def generate_geohash_column(df, precision=8):
    if 'latitude' in df.columns and 'longitude' in df.columns:
        print("Generating geohash from latitude and longitude...")
        df['geohash'] = df.apply(lambda row: geohash2.encode(row['latitude'], row['longitude'], precision=precision), axis=1)
    elif 'geometry' in df.columns and df['geometry'].notna().all():
        print("Extracting coordinates from geometry and generating geohash...")
        df['geohash'] = df['geometry'].apply(lambda geom: geohash2.encode(geom.centroid.y, geom.centroid.x, precision=precision) if geom else None)
    else:
        print("No coordinate data available to generate geohash.")
    return df

# Apply geohash generation
cleaned_df = generate_geohash_column(cleaned_df)

# Convert to Table again
final_table = Table.from_df(cleaned_df)

print("Displaying the final table with geohash:")
final_table.show(3)
```

Now let’s add code that downloads the final_table as a CSV file directly from Colab.

Since final_table is a datascience.Table, we first convert it back to a pandas DataFrame, save it as a CSV, then use Colab's files.download() to prompt a download.

```python
# --- Export and download the final table as a CSV file ---
from google.colab import files

# Convert Table to DataFrame
final_df = final_table.to_df()

# Save to CSV
csv_filename = "final_table.csv"
final_df.to_csv(csv_filename, index=False)

# Prompt download
files.download(csv_filename)
```
